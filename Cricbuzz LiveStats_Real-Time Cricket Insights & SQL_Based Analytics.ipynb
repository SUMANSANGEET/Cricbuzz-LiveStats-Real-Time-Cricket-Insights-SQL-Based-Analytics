# 🏏 Project Summary

**Cricbuzz LiveStats: Real-Time Cricket Insights & SQL-Based Analytics** is an end-to-end **Sports Data Analytics project** that integrates live cricket information from the **Cricbuzz API via REST API**, processes the data using Python, stores structured information in a relational SQL database, and presents actionable insights through interactive analytics and visualization.

The project is designed to create a centralized platform for analyzing **matches, players, teams, venues, performance statistics, and historical cricket trends**. Python is used as the primary data-processing and API-integration language, while SQL provides the foundation for structured data storage, querying, aggregation, and advanced analytics.

The solution follows an ETL-oriented workflow:

**API → Data Extraction → Data Cleaning → Data Validation → SQL Database → Analytics → Visualization → Streamlit Dashboard**

The analytical component contains **25 SQL-based business questions**, progressing from beginner-level filtering and aggregation to intermediate JOINs and subqueries and advanced analytical techniques involving **CTEs, window functions, and complex calculations**. These analyses cover player performance, team wins, venue statistics, toss impact, batting partnerships, home-versus-away performance, player consistency, recent form, and performance evolution.

Interactive visualizations are used to convert analytical results into understandable business insights. The project incorporates KPI summaries, player comparisons, team-performance charts, trend analysis, venue comparisons, and other interactive visual exploration techniques.

The project also includes a **CRUD-based data management module**, allowing authorized users to create, read, update, and delete player and match-related records through a form-based interface.

The final application is planned as a **multi-page Streamlit dashboard** containing modules for:

* ⚡ Live Match Updates
* 🏏 Top Player Statistics
* 📊 Cricket Analytics
* 🧮 SQL Query Analytics
* ✏️ CRUD Data Management
* 🏠 Project Overview and Documentation

The project emphasizes professional development practices including **PEP 8 coding standards, modular programming, exception handling, secure API/database credentials, data validation, documentation, and Git-based version control**.

### Project Outcome

The completed solution will demonstrate an end-to-end data workflow in which raw cricket API data is transformed into **structured, queryable, interactive, and decision-oriented cricket intelligence**. It provides practical evidence of skills in:

**Python • SQL • REST APIs • JSON • Pandas • Data Analytics • Data Visualization • Database Management • Streamlit • CRUD • Sports Analytics**

# 🎯 Problem Statement

Cricket generates a large volume of data across **matches, players, teams, venues, series, batting performances, bowling performances, and match outcomes**. However, raw cricket data obtained through APIs is typically delivered in structured JSON responses that are not immediately suitable for efficient querying, comparative analysis, visualization, or business decision-making.

The objective of this project is to develop **Cricbuzz LiveStats**, an end-to-end cricket analytics platform that integrates live cricket data from the **Cricbuzz API through RapidAPI**, processes and validates the incoming data using Python, and stores structured information in a relational SQL database.

The system addresses the challenge of converting raw and continuously changing cricket data into a **centralized, reliable, and interactive analytical environment**. It enables users to monitor live matches, explore player statistics, analyze team and venue performance, investigate historical trends, and answer business-oriented questions using SQL.

A major challenge is that meaningful cricket analysis requires more than basic filtering or visualization. Analysts need to evaluate relationships such as:

* Which players are performing consistently?
* Which teams have the strongest win records?
* Does winning the toss provide an advantage?
* How do teams perform at home versus away?
* Which venues favor particular players or bowling styles?
* Which batting partnerships are most successful?
* How has a player's performance changed over time?
* Which players are currently demonstrating strong recent form?

To address these analytical requirements, the platform implements **25 SQL-based analytical questions**, progressing from basic SQL operations to advanced techniques such as **JOINs, subqueries, aggregations, Common Table Expressions (CTEs), window functions, and analytical calculations**.

The project further addresses the need for accessible data interpretation by converting analytical outputs into **interactive visualizations and dashboards**. Instead of relying solely on raw tables, users can explore player, team, venue, match, and performance trends through interactive charts.

The system also incorporates a **CRUD data-management module**, allowing authorized users to create, read, update, and delete relevant cricket records through a controlled interface.

## Q21. Rank players by total runs using a window function

### Business Question

> How do players rank based on their total run contribution?

```sql
WITH player_runs AS (

        p.player_id,

SELECT
    player_id,
    player_name,
    total_runs,
    RANK() OVER (
        ORDER BY total_runs DESC

FROM player_runs
ORDER BY run_rank;
```

````python
USE_LIVE_API = True  # flip to True once RAPIDAPI_KEY is set as an env var
RAPIDAPI_KEY = os.environ.get("RAPIDAPI_KEY", "")
RAPIDAPI_HOST = "cricbuzz-cricket.p.rapidapi.com"
BASE_URL = f"https://{RAPIDAPI_HOST}"

HEADERS = {
    "X-RapidAPI-Key": RAPIDAPI_KEY,
    "X-RapidAPI-Host": RAPIDAPI_HOST,
}

class CricbuzzAPIError(Exception):
    pass

def fetch_endpoint(path: str, params: dict = None, max_retries: int = 3, backoff: float = 1.5):
    """Fetch a Cricbuzz endpoint with retry/backoff. Falls back to local fixtures when
    USE_LIVE_API is False or the network call fails, so downstream code never breaks."""
    if not USE_LIVE_API:
        return _local_fixture(path)

    import requests
    for attempt in range(1, max_retries + 1):
        try:
            resp = requests.get(f"{BASE_URL}{path}", headers=HEADERS, params=params, timeout=10)
            if resp.status_code == 200:
                return resp.json()
            if resp.status_code == 429:
                time.sleep(backoff ** attempt)
                continue
            raise CricbuzzAPIError(f"{resp.status_code}: {resp.text[:200]}")
        except requests.exceptions.RequestException as e:
            if attempt == max_retries:
                print(f"API call failed after {max_retries} attempts ({e}); using local fixture.")
                return _local_fixture(path)
            time.sleep(backoff ** attempt)

def _local_fixture(path: str):
    return {"fixture_for": path, "note": "served by synthetic data generator, see Section 3"}

print("API layer configured. USE_LIVE_API =", USE_LIVE_API)
````

### Live endpoint — full match scorecard (`hscard`)

Wraps `GET /mcenter/v1/{match_id}/hscard`, the endpoint that returns every innings' batting and bowling scorecard for one match. It reuses `fetch_endpoint()` above, so it automatically serves synthetic fixtures whenever `USE_LIVE_API = False` — no branching logic duplicated here.

````python
def get_full_scorecard(match_id: int) -> dict:
    """GET /mcenter/v1/{match_id}/hscard via the shared retry/backoff client (Section 2)."""
    return fetch_endpoint(f"/mcenter/v1/{match_id}/hscard")


def scorecard_to_dataframes(raw: dict, match_id: int):
    """Parses an hscard payload into batting_scorecards / bowling_scorecards rows.

    Cricbuzz's internal API isn't officially documented and key names have shifted
    across app versions, so this is written defensively (.get() everywhere) and is
    meant to be checked against one real payload before trusting it at scale — see
    the raw-print cell below.
    """
    bat_rows, bowl_rows = [], []
    for innings in raw.get("scoreCard", []):
        bat_team = innings.get("batTeamDetails", {})
        bat_team_id = bat_team.get("batTeamId")
        for b in bat_team.get("batsmenData", {}).values():
            bat_rows.append({
                "match_id": match_id,
                "player_id": b.get("batId"),
                "team_id": bat_team_id,
                "runs": b.get("runs"),
                "balls_faced": b.get("balls"),
                "fours": b.get("fours"),
                "sixes": b.get("sixes"),
                "strike_rate": b.get("strikeRate"),
                "is_out": 0 if b.get("outDesc", "").lower() == "not out" else 1,
            })

        bowl_team = innings.get("bowlTeamDetails", {})
        bowl_team_id = bowl_team.get("bowlTeamId")
        for bo in bowl_team.get("bowlersData", {}).values():
            bowl_rows.append({
                "match_id": match_id,
                "player_id": bo.get("bowlerId"),
                "team_id": bowl_team_id,
                "overs": bo.get("overs"),
                "runs_conceded": bo.get("runs"),
                "wickets": bo.get("wickets"),
                "economy": bo.get("economy"),
                "maidens": bo.get("maidens"),
            })

    return pd.DataFrame(bat_rows), pd.DataFrame(bowl_rows)


print("Live scorecard helpers ready: get_full_scorecard(), scorecard_to_dataframes()")
````

## 3 ·🧪 Synthetic Data Generator (local fallback / dev fixtures)

Generates a realistic multi-series cricket universe: teams, players, venues, matches, and ball-by-ball-derived batting/bowling scorecards — matching the shape the real API would return, so the schema and queries downstream are unaffected by the data source.

````python
TEAMS = ["India", "Australia", "England", "New Zealand", "Pakistan",
         "South Africa", "Sri Lanka", "West Indies", "Bangladesh", "Afghanistan"]

FIRST_NAMES = ["Rohit","Virat","Steve","Joe","Kane","Babar","Quinton","Shai","Shakib","Rashid",
               "Jasprit","Pat","Trent","Mitchell","Fakhar","David","Ben","Marnus","Rassie","Mushfiqur",
               "Suryakumar","Glenn","Jos","Devon","Shaheen","Kagiso","Wanindu","Rishabh","Travis","Hardik"]
LAST_NAMES = ["Sharma","Kohli","Smith","Root","Williamson","Azam","de Kock","Hope","Al Hasan","Khan",
              "Bumrah","Cummins","Boult","Starc","Zaman","Warner","Stokes","Labuschagne","van der Dussen","Rahim",
              "Yadav","Maxwell","Buttler","Conway","Afridi","Rabada","Fernando","Pant","Head","Pandya"]

ROLES = ["Batsman","Bowler","All-Rounder","Wicket-Keeper"]
FORMATS = ["T20I","ODI","Test"]
VENUE_CITIES = ["Mumbai","Melbourne","London","Auckland","Lahore","Johannesburg","Colombo",
                "Bridgetown","Dhaka","Kabul","Chennai","Sydney","Manchester","Wellington","Karachi"]

def build_teams():
    return pd.DataFrame({"team_id": range(1, len(TEAMS)+1), "team_name": TEAMS,
                          "region": np.random.choice(["Asia","Oceania","Europe","Africa","Americas"], len(TEAMS))})

def build_players(teams_df, n_per_team=15):
    rows, pid = [], 1
    for _, t in teams_df.iterrows():
        for _ in range(n_per_team):
            rows.append({
                "player_id": pid,
                "player_name": f"{random.choice(FIRST_NAMES)} {random.choice(LAST_NAMES)}",
                "team_id": t.team_id,
                "role": np.random.choice(ROLES, p=[0.4,0.3,0.2,0.1]),
                "batting_style": np.random.choice(["Right-hand bat","Left-hand bat"], p=[0.75,0.25]),
                "bowling_style": np.random.choice(["Right-arm fast","Left-arm fast","Right-arm off-spin",
                                                     "Left-arm orthodox","Leg-spin","None"], p=[0.25,0.15,0.2,0.1,0.15,0.15]),
                "dob": (datetime(1988,1,1) + timedelta(days=int(np.random.uniform(0, 365*15)))).date(),
            })
            pid += 1
    return pd.DataFrame(rows)

def build_venues(n=15):
    return pd.DataFrame({
        "venue_id": range(1, n+1),
        "venue_name": [f"{c} International Stadium" for c in VENUE_CITIES[:n]],
        "city": VENUE_CITIES[:n],
        "country": np.random.choice(["India","Australia","England","NZ","Pakistan","SA","SL","WI","Bangladesh","Afghanistan"], n),
        "capacity": np.random.randint(15000, 100000, n),
    })

def build_series(n=6):
    starts = [datetime(2022,1,1) + timedelta(days=int(d)) for d in np.random.uniform(0, 900, n)]
    return pd.DataFrame({
        "series_id": range(1, n+1),
        "series_name": [f"{random.choice(TEAMS)} tour of {random.choice(TEAMS)} {s.year}" for s in starts],
        "format": np.random.choice(FORMATS, n),
        "start_date": [s.date() for s in starts],
    })

def build_matches(teams_df, venues_df, series_df, n=120):
    rows = []
    for mid in range(1, n+1):
        t1, t2 = np.random.choice(teams_df.team_id, 2, replace=False)
        series = series_df.sample(1).iloc[0]
        venue = venues_df.sample(1).iloc[0]
        m_date = series.start_date + timedelta(days=int(np.random.uniform(0,25)))
        toss_winner = np.random.choice([t1, t2])
        toss_decision = np.random.choice(["Bat","Field"])
        margin_type = np.random.choice(["runs","wickets"])
        margin = np.random.randint(1, 120) if margin_type == "runs" else np.random.randint(1,10)
        winner = np.random.choice([t1, t2, None], p=[0.47,0.47,0.06])  # 6% no-result/tie
        rows.append({
            "match_id": mid, "series_id": series.series_id, "format": series.format,
            "team1_id": t1, "team2_id": t2, "venue_id": venue.venue_id, "match_date": m_date,
            "toss_winner_id": toss_winner, "toss_decision": toss_decision,
            "winner_team_id": winner, "win_margin": margin if winner is not None else None,
            "win_margin_type": margin_type if winner is not None else None,
            "player_of_match_id": None,  # filled after scorecards are generated
        })
    return pd.DataFrame(rows)

def build_scorecards(matches_df, players_df):
    """Generate batting + bowling innings for every match, then back-fill Player-of-the-Match."""
    bat_rows, bowl_rows = [], []
    bat_id = bowl_id = 1
    pom_by_match = {}

    for _, m in matches_df.iterrows():
        for team_id in [m.team1_id, m.team2_id]:
            squad = players_df[players_df.team_id == team_id].sample(11)
            batters = squad[squad.role != "Bowler"].sample(min(8, len(squad[squad.role != "Bowler"])))
            best_score, best_player = -1, None
            for _, p in batters.iterrows():
                runs = int(np.random.gamma(2.2, 18))
                balls = max(runs, int(runs / np.random.uniform(0.55, 1.3)) + np.random.randint(0,10))
                fours = int(np.random.poisson(runs/12))
                sixes = int(np.random.poisson(runs/30))
                bat_rows.append({
                    "scorecard_id": bat_id, "match_id": m.match_id, "player_id": p.player_id,
                    "team_id": team_id, "runs": runs, "balls_faced": max(balls,1), "fours": fours,
                    "sixes": sixes, "strike_rate": round(runs / max(balls,1) * 100, 2),
                    "is_out": bool(np.random.choice([True, False], p=[0.82, 0.18])),
                })
                impact = runs + fours*1 + sixes*2
                if impact > best_score:
                    best_score, best_player = impact, p.player_id
                bat_id += 1

            bowlers = squad[squad.role.isin(["Bowler","All-Rounder"])]
            if len(bowlers) == 0:
                bowlers = squad.sample(3)
            for _, p in bowlers.iterrows():
                overs = round(np.random.uniform(2,10) * 2)/2
                economy = round(np.random.uniform(3.5, 10.5), 2)
                runs_conceded = int(overs * economy)
                wkts = int(np.random.choice([0,1,2,3,4,5], p=[0.30,0.28,0.20,0.12,0.07,0.03]))
                bowl_rows.append({
                    "bowling_id": bowl_id, "match_id": m.match_id, "player_id": p.player_id,
                    "team_id": team_id, "overs": overs, "runs_conceded": runs_conceded,
                    "wickets": wkts, "economy": economy, "maidens": int(np.random.poisson(0.3)),
                })
                bowl_id += 1
            if random.random() < 0.5 and best_player is not None:
                pom_by_match[m.match_id] = best_player
        if m.match_id not in pom_by_match:
            pom_by_match[m.match_id] = players_df[players_df.team_id.isin([m.team1_id,m.team2_id])].sample(1).iloc[0].player_id

    matches_df = matches_df.copy()
    matches_df["player_of_match_id"] = matches_df.match_id.map(pom_by_match)
    return pd.DataFrame(bat_rows), pd.DataFrame(bowl_rows), matches_df

teams_df = build_teams()
players_df = build_players(teams_df)
venues_df = build_venues()
series_df = build_series()
matches_df = build_matches(teams_df, venues_df, series_df)
batting_df, bowling_df, matches_df = build_scorecards(matches_df, players_df)

print(f"teams={len(teams_df)}  players={len(players_df)}  venues={len(venues_df)}  "
      f"series={len(series_df)}  matches={len(matches_df)}  batting_rows={len(batting_df)}  bowling_rows={len(bowling_df)}")
matches_df.head()
````

## 4 ·🗄️ Database Design — SQLite Schema (8 normalized tables)

**Entities:** `teams`, `players`, `venues`, `series`, `matches`, `batting_scorecards`, `bowling_scorecards`, plus a derived `partnerships` view.
Normalized to 3NF — matches reference teams/venues/series by FK, scorecards reference matches/players by FK. Indexes on every FK for join performance.

````python
conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()

cur.executescript('''
DROP TABLE IF EXISTS bowling_scorecards;
DROP TABLE IF EXISTS batting_scorecards;
DROP TABLE IF EXISTS matches;
DROP TABLE IF EXISTS series;
DROP TABLE IF EXISTS venues;
DROP TABLE IF EXISTS players;
DROP TABLE IF EXISTS teams;

CREATE TABLE teams (
    team_id INTEGER PRIMARY KEY,
    team_name TEXT NOT NULL UNIQUE,
    region TEXT
);

CREATE TABLE players (
    player_id INTEGER PRIMARY KEY,
    player_name TEXT NOT NULL,
    team_id INTEGER NOT NULL,
    role TEXT,
    batting_style TEXT,
    bowling_style TEXT,
    dob DATE,
    FOREIGN KEY (team_id) REFERENCES teams(team_id)
);

CREATE TABLE venues (
    venue_id INTEGER PRIMARY KEY,
    venue_name TEXT NOT NULL,
    city TEXT,
    country TEXT,
    capacity INTEGER
);

CREATE TABLE series (
    series_id INTEGER PRIMARY KEY,
    series_name TEXT NOT NULL,
    format TEXT CHECK(format IN ('T20I','ODI','Test')),
    start_date DATE
);

CREATE TABLE matches (
    match_id INTEGER PRIMARY KEY,
    series_id INTEGER NOT NULL,
    format TEXT,
    team1_id INTEGER NOT NULL,
    team2_id INTEGER NOT NULL,
    venue_id INTEGER NOT NULL,
    match_date DATE,
    toss_winner_id INTEGER,
    toss_decision TEXT,
    winner_team_id INTEGER,
    win_margin INTEGER,
    win_margin_type TEXT,
    player_of_match_id INTEGER,
    FOREIGN KEY (series_id) REFERENCES series(series_id),
    FOREIGN KEY (team1_id) REFERENCES teams(team_id),
    FOREIGN KEY (team2_id) REFERENCES teams(team_id),
    FOREIGN KEY (venue_id) REFERENCES venues(venue_id),
    FOREIGN KEY (winner_team_id) REFERENCES teams(team_id),
    FOREIGN KEY (player_of_match_id) REFERENCES players(player_id)
);

CREATE TABLE batting_scorecards (
    scorecard_id INTEGER PRIMARY KEY,
    match_id INTEGER NOT NULL,
    player_id INTEGER NOT NULL,
    team_id INTEGER NOT NULL,
    runs INTEGER, balls_faced INTEGER, fours INTEGER, sixes INTEGER,
    strike_rate REAL, is_out INTEGER,
    FOREIGN KEY (match_id) REFERENCES matches(match_id),
    FOREIGN KEY (player_id) REFERENCES players(player_id)
);

CREATE TABLE bowling_scorecards (
    bowling_id INTEGER PRIMARY KEY,
    match_id INTEGER NOT NULL,
    player_id INTEGER NOT NULL,
    team_id INTEGER NOT NULL,
    overs REAL, runs_conceded INTEGER, wickets INTEGER, economy REAL, maidens INTEGER,
    FOREIGN KEY (match_id) REFERENCES matches(match_id),
    FOREIGN KEY (player_id) REFERENCES players(player_id)
);

CREATE INDEX idx_players_team ON players(team_id);
CREATE INDEX idx_matches_series ON matches(series_id);
CREATE INDEX idx_matches_venue ON matches(venue_id);
CREATE INDEX idx_bat_match ON batting_scorecards(match_id);
CREATE INDEX idx_bat_player ON batting_scorecards(player_id);
CREATE INDEX idx_bowl_match ON bowling_scorecards(match_id);
CREATE INDEX idx_bowl_player ON bowling_scorecards(player_id);
''')
conn.commit()
print("Schema created: 6 base tables + 7 indexes.")
````

**Q13. Rank the top run scorer within each team (`RANK()` window function).**

````python
q13 = run_sql('''
WITH team_runs AS (
  SELECT p.team_id, p.player_id, p.player_name, SUM(b.runs) AS total_runs
  FROM batting_scorecards b JOIN players p ON p.player_id=b.player_id
  GROUP BY p.player_id
)
SELECT t.team_name, tr.player_name, tr.total_runs,
       RANK() OVER (PARTITION BY tr.team_id ORDER BY tr.total_runs DESC) AS team_rank
FROM team_runs tr JOIN teams t ON t.team_id = tr.team_id
QUALIFY team_rank = 1
ORDER BY total_runs DESC;
''') if False else None

# SQLite (pre-3.44) has no QUALIFY -> filter with a subquery instead
q13 = run_sql('''
WITH team_runs AS (
  SELECT p.team_id, p.player_name, SUM(b.runs) AS total_runs,
         RANK() OVER (PARTITION BY p.team_id ORDER BY SUM(b.runs) DESC) AS team_rank
  FROM batting_scorecards b JOIN players p ON p.player_id=b.player_id
  GROUP BY p.player_id
)
SELECT t.team_name, tr.player_name, tr.total_runs
FROM team_runs tr JOIN teams t ON t.team_id = tr.team_id
WHERE tr.team_rank = 1
ORDER BY tr.total_runs DESC;
''')
q13
````

**Q16. Consistency index — coefficient of variation of runs per player (lower = more consistent), min. 8 innings.**

````python
q16 = run_sql('''
SELECT
    p.player_id,
    p.player_name,
    t.team_name,
    COUNT(*) AS innings,
    ROUND(AVG(b.runs), 1) AS avg_runs
FROM batting_scorecards b
JOIN players p
    ON p.player_id = b.player_id
JOIN teams t
    ON t.team_id = p.team_id
GROUP BY
    p.player_id,
    p.player_name,
    t.team_name
HAVING COUNT(*) >= 8
''')

q16["stdev_runs"] = q16.apply(
    lambda row: (
        (
            run_sql(f'''
                SELECT AVG(
                    (runs - {row["avg_runs"]}) *
                    (runs - {row["avg_runs"]})
                ) AS variance
                FROM batting_scorecards
                WHERE player_id = {int(row["player_id"])}
            ''').iloc[0]["variance"]
        ) ** 0.5
    ),
    axis=1
)

q16["stdev_runs"] = q16["stdev_runs"].round(1)

q16 = (
    q16[
        ["player_name", "team_name", "innings",
         "avg_runs", "stdev_runs"]
    ]
    .sort_values("stdev_runs")
    .head(10)
    .reset_index(drop=True)
)

q16
````

**Q19. Most consistent batsmen — average runs & standard deviation (min. 10 qualifying innings, ≥10 balls faced, since 2022).** Lower std. dev. = more consistent.

````python
import math

conn.create_function("SQRT", 1, math.sqrt)

print("✅ SQRT() function registered successfully")
````

````python
q19 = run_sql('''
WITH filtered AS (
  SELECT b.player_id, b.runs
  FROM batting_scorecards b
  JOIN matches m ON m.match_id = b.match_id
  WHERE b.balls_faced >= 10 AND m.match_date >= \'2022-01-01\'
),
stats AS (
  SELECT player_id, COUNT(*) AS innings, AVG(runs) AS avg_runs
  FROM filtered GROUP BY player_id HAVING COUNT(*) >= 10
)
SELECT p.player_name, t.team_name, s.innings, ROUND(s.avg_runs,1) AS avg_runs,
       ROUND(SQRT(AVG((f.runs - s.avg_runs)*(f.runs - s.avg_runs))),1) AS stdev_runs
FROM stats s
JOIN filtered f ON f.player_id = s.player_id
JOIN players p ON p.player_id = s.player_id
JOIN teams t ON t.team_id = p.team_id
GROUP BY s.player_id
ORDER BY stdev_runs ASC
LIMIT 10;
''')
q19
````

**Q21. Composite performance ranking — brief's weighted formula, ranked within each format.**

`batting_points = runs×0.01 + batting_avg×0.5 + strike_rate×0.3`
`bowling_points = wickets×2 + (50 − bowling_avg)×0.5 + (6 − economy)×2`

The brief also weights a fielding component (`catches×3 + stumpings×5`); the Cricbuzz scorecard fields modeled in this schema (Section 4) don't carry fielding dismissals, so that term is omitted here and called out explicitly rather than silently dropped — a production build would add a `fielding_stats` table and reinstate it unchanged.

````python
q21 = run_sql('''
WITH bat AS (
  SELECT b.player_id, m.format, SUM(b.runs) AS runs,
         SUM(b.runs)*1.0/NULLIF(SUM(CASE WHEN b.is_out=1 THEN 1 ELSE 0 END),0) AS batting_avg,
         AVG(b.strike_rate) AS strike_rate
  FROM batting_scorecards b JOIN matches m ON m.match_id=b.match_id
  GROUP BY b.player_id, m.format
),
bowl AS (
  SELECT bo.player_id, m.format, SUM(bo.wickets) AS wickets,
         SUM(bo.runs_conceded)*1.0/NULLIF(SUM(bo.wickets),0) AS bowling_avg,
         SUM(bo.runs_conceded)*1.0/NULLIF(SUM(bo.overs),0) AS economy
  FROM bowling_scorecards bo JOIN matches m ON m.match_id=bo.match_id
  GROUP BY bo.player_id, m.format
),
scored AS (
  SELECT p.player_name, t.team_name, bat.format,
    ROUND(COALESCE(bat.runs,0)*0.01 + COALESCE(bat.batting_avg,0)*0.5 + COALESCE(bat.strike_rate,0)*0.3,1) AS batting_points,
    ROUND(COALESCE(bowl.wickets,0)*2 + (50-COALESCE(bowl.bowling_avg,50))*0.5 + (6-COALESCE(bowl.economy,6))*2,1) AS bowling_points
  FROM bat
  LEFT JOIN bowl ON bowl.player_id=bat.player_id AND bowl.format=bat.format
  JOIN players p ON p.player_id=bat.player_id
  JOIN teams t ON t.team_id=p.team_id
)
SELECT *, ROUND(batting_points+bowling_points,1) AS total_points,
  RANK() OVER (PARTITION BY format ORDER BY (batting_points+bowling_points) DESC) AS format_rank
FROM scored
ORDER BY format, format_rank;
''')
q21_top = q21[q21.format_rank <= 5].reset_index(drop=True)
q21_top
````

````python
fig = px.bar(q21_top, x="total_points", y="player_name", color="format", orientation="h",
             title="Q21 · Composite Performance Ranking — Top 5 per Format",
             labels={"total_points":"Weighted Score","player_name":"Player"}, facet_col="format")
fig.update_yaxes(matches=None, showticklabels=True)
fig.update_layout(height=450)
fig.show()
````

**Q22. Head-to-head prediction analysis — team pairs with ≥5 meetings: win split and average victory margin.**

**Q23. Recent form & momentum — last 10 innings per player: recency-weighted average, strike-rate trend, and a form label.**

````python
q23 = run_sql('''
WITH ranked AS (
  SELECT b.player_id, b.runs, b.strike_rate, m.match_date,
     ROW_NUMBER() OVER (PARTITION BY b.player_id ORDER BY m.match_date DESC) AS rn
  FROM batting_scorecards b JOIN matches m ON m.match_id=b.match_id
),
last10 AS (SELECT * FROM ranked WHERE rn<=10)
SELECT p.player_name, t.team_name,
  ROUND(AVG(CASE WHEN l10.rn<=5 THEN l10.runs END),1) AS avg_last5,
  ROUND(AVG(l10.runs),1) AS avg_last10,
  ROUND(AVG(l10.strike_rate),1) AS sr_last10,
  SUM(CASE WHEN l10.runs>=50 THEN 1 ELSE 0 END) AS scores_50plus,
  COUNT(*) AS innings_counted
FROM last10 l10
JOIN players p ON p.player_id=l10.player_id
JOIN teams t ON t.team_id=p.team_id
GROUP BY l10.player_id
HAVING innings_counted = 10
ORDER BY avg_last10 DESC
LIMIT 15;
''')

def form_label(row):
    if row.avg_last10 >= 40 and row.avg_last5 >= row.avg_last10:
        return "Excellent Form"
    if row.avg_last10 >= 30:
        return "Good Form"
    if row.avg_last10 >= 18:
        return "Average Form"
    return "Poor Form"

q23["form"] = q23.apply(form_label, axis=1)
q23
````

**Q24. Best batting-partnership combinations — pairs with ≥5 shared innings, ranked by success rate (share of 50+ partnerships).**

````python
q24 = run_sql('''
WITH partnerships AS (
  SELECT b1.match_id, b1.team_id, b1.player_id AS p1, b2.player_id AS p2,
         (b1.runs+b2.runs) AS combined_runs
  FROM batting_scorecards b1
  JOIN batting_scorecards b2 ON b1.match_id=b2.match_id AND b1.team_id=b2.team_id AND b1.player_id<b2.player_id
),
agg AS (
  SELECT p1,p2, COUNT(*) AS partnership_count, AVG(combined_runs) AS avg_runs,
         MAX(combined_runs) AS best_runs,
         SUM(CASE WHEN combined_runs>=50 THEN 1 ELSE 0 END) AS good_partnerships
  FROM partnerships GROUP BY p1,p2 HAVING COUNT(*)>=5
)
SELECT pl1.player_name AS batter_1, pl2.player_name AS batter_2,
  a.partnership_count, ROUND(a.avg_runs,1) AS avg_partnership_runs,
  a.best_runs AS highest_partnership,
  ROUND(100.0*a.good_partnerships/a.partnership_count,1) AS success_rate_pct
FROM agg a
JOIN players pl1 ON pl1.player_id=a.p1
JOIN players pl2 ON pl2.player_id=a.p2
ORDER BY success_rate_pct DESC, avg_partnership_runs DESC
LIMIT 10;
''')
q24
````

**Q25. Career trajectory — quarterly batting evolution, classified as Ascending / Declining / Stable.**

Brief's thresholds are "≥6 qualifying quarters, ≥3 matches per quarter"; on this demo dataset that's relaxed to **≥4 quarters / ≥2 matches per quarter** so the trend logic below has enough qualifying players to demonstrate on — the query and trend classifier are otherwise exactly what a live-API-scale dataset would run unchanged.

````python
q25 = run_sql('''
WITH quarterly AS (
  SELECT b.player_id,
    (CAST(strftime(\'%Y\', m.match_date) AS INT)*4 + (CAST(strftime(\'%m\', m.match_date) AS INT)-1)/3) AS q_index,
    strftime(\'%Y\', m.match_date) || \'-Q\' || ((CAST(strftime(\'%m\', m.match_date) AS INT)-1)/3 + 1) AS quarter_label,
    b.runs, b.strike_rate
  FROM batting_scorecards b JOIN matches m ON m.match_id=b.match_id
),
qagg AS (
  SELECT player_id, q_index, quarter_label, COUNT(*) AS matches_in_q, AVG(runs) AS avg_runs, AVG(strike_rate) AS avg_sr
  FROM quarterly GROUP BY player_id, q_index HAVING COUNT(*)>=2
),
qualified AS (
  SELECT player_id FROM qagg GROUP BY player_id HAVING COUNT(*)>=4
)
SELECT p.player_name, t.team_name, q.q_index, q.quarter_label, ROUND(q.avg_runs,1) AS avg_runs, ROUND(q.avg_sr,1) AS avg_sr
FROM qagg q
JOIN qualified ql ON ql.player_id=q.player_id
JOIN players p ON p.player_id=q.player_id
JOIN teams t ON t.team_id=p.team_id
ORDER BY p.player_name, q.q_index;
''')

def classify_trajectory(group):
    x = np.arange(len(group))
    slope = np.polyfit(x, group["avg_runs"], 1)[0] if len(group) >= 2 else 0
    if slope > 1.5:
        return "Career Ascending"
    if slope < -1.5:
        return "Career Declining"
    return "Career Stable"

trajectories = (q25.groupby("player_name", group_keys=False)
                    .apply(lambda g: pd.Series({"team_name": g.team_name.iloc[0],
                                                 "quarters": len(g),
                                                 "trajectory": classify_trajectory(g)}))
                    .reset_index())
trajectories.trajectory.value_counts()
````

````python
sample_players = trajectories.sort_values("quarters", ascending=False).player_name.head(5).tolist()
plot_df = q25[q25.player_name.isin(sample_players)]
fig = px.line(plot_df, x="quarter_label", y="avg_runs", color="player_name", markers=True,
              title="Q25 · Quarterly Batting Trajectory — Top-Coverage Players",
              labels={"quarter_label":"Quarter","avg_runs":"Avg Runs"})
fig.update_layout(height=450)
fig.show()
````

## 8 · 🔒 CRUD Module (with transaction safety)

Production-style insert / update / delete against `matches`, wrapped in explicit transactions with rollback on failure.m

````python
def create_match(record: dict) -> int:
    with conn:
        cur = conn.execute('''INSERT INTO matches
            (series_id, format, team1_id, team2_id, venue_id, match_date, toss_winner_id,
             toss_decision, winner_team_id, win_margin, win_margin_type, player_of_match_id)
            VALUES (:series_id,:format,:team1_id,:team2_id,:venue_id,:match_date,:toss_winner_id,
                    :toss_decision,:winner_team_id,:win_margin,:win_margin_type,:player_of_match_id)''', record)
        return cur.lastrowid

def read_match(match_id: int) -> pd.DataFrame:
    return run_sql(f"SELECT * FROM matches WHERE match_id = {match_id}")

def update_match(match_id: int, **fields) -> int:
    set_clause = ", ".join(f"{k} = :{k}" for k in fields)
    fields["match_id"] = match_id
    with conn:
        cur = conn.execute(f"UPDATE matches SET {set_clause} WHERE match_id = :match_id", fields)
        return cur.rowcount

def delete_match(match_id: int) -> int:
    with conn:
        cur = conn.execute("DELETE FROM matches WHERE match_id = ?", (match_id,))
        return cur.rowcount

# --- demo ---
new_id = create_match({
    "series_id": 1, "format": "T20I", "team1_id": 1, "team2_id": 2, "venue_id": 1,
    "match_date": "2026-09-04", "toss_winner_id": 1, "toss_decision": "Bat",
    "winner_team_id": 1, "win_margin": 18, "win_margin_type": "runs", "player_of_match_id": 1
})
print("Inserted match_id:", new_id)
print(read_match(new_id).to_string(index=False))

update_match(new_id, win_margin=25)
print("\nAfter UPDATE:")
print(read_match(new_id)[["match_id","win_margin"]].to_string(index=False))

rows_deleted = delete_match(new_id)
print(f"\nDeleted {rows_deleted} row(s). CRUD cycle verified end-to-end.")
````
